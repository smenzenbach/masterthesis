```python
from PIL import Image
import requests
from transformers import SamModel, SamProcessor, pipeline
from llama_index.core import SimpleDirectoryReader
import os
import numpy as np
import matplotlib.pyplot as plt
import gc
```

```python
os.environ["CUDA_DEVICE_ORDER"] = "PCI_BUS_ID"
os.environ["CUDA_VISIBLE_DEVICES"] = "2"
model = SamModel.from_pretrained("facebook/sam-vit-large")
processor = SamProcessor.from_pretrained("facebook/sam-vit-large")
```

```python
def metadata_str2dict(string):
    # Split the string into lines
    lines = string.strip().split("\n")
    
    # Create a dictionary to store the key-value pairs
    file_info_dict = {}
    
    # Parse each line and add key-value pairs to the dictionary
    for line in lines:
        key, value = line.split(": ", 1)
        file_info_dict[key] = value
    
    return(file_info_dict)
```

```python
# load as image documents
image_documents = SimpleDirectoryReader("/home/data/ISIC2019/raw", required_exts=[".jpeg", ".jpg", ".png"], recursive=True, filename_as_id=True).load_data()

image_patch = []
image_no_patch = []

for image in image_documents:
    metadata = metadata_str2dict(image.get_metadata_str())
    image_path = metadata['file_path']

    folder_name = os.path.basename(os.path.dirname(image_path))
    
    if folder_name == "patch":
        metadata['patch'] = True
        image_patch.append({'image': image, 'meta': metadata})


    elif folder_name == "no_patch":
        metadata['patch'] = False
        image_no_patch.append({'image': image, 'meta': metadata})


    else:
        print("The jpg is not in either the patch or no_patch folder.")
```

```python
def show_mask(mask, ax, random_color=False):
    if random_color:
        color = np.concatenate([np.random.random(3), np.array([0.6])], axis=0)
    else:
        color = np.array([30 / 255, 144 / 255, 255 / 255, 0.6])
    h, w = mask.shape[-2:]
    mask_image = mask.reshape(h, w, 1) * color.reshape(1, 1, -1)
    ax.imshow(mask_image)
    del mask
    gc.collect()

def masks_on_image(raw_image, masks, show=True, save=False, folderpath=None, filename=None):
  plt.imshow(np.array(raw_image))
  ax = plt.gca()
  ax.set_autoscale_on(False)
  seed=42
  for mask in masks:
      np.random.seed(seed)
      show_mask(mask, ax=ax, random_color=True)
      seed += 7
  plt.axis("off")
  if show:
      plt.show()

  if save:
      if folderpath is None:
          raise TypeError("Please provide folderpath")
      if filename is None:
          raise TypeError("Please provide filename")
          
      # Ensure the folder exists
      if not os.path.exists(folderpath):
          os.makedirs(folderpath)
    
      # Save the figure
      save_path = os.path.join(folderpath, filename)
      plt.savefig(save_path, format='jpg', bbox_inches='tight', pad_inches=0)
      print(f"save img {filename}")
      if not show:
          plt.close()
  del mask
  gc.collect()
```

```python
meta = image_patch[0]['meta']
imageUrl = meta['file_path']

generator =  pipeline("mask-generation", model=model, image_processor=processor.image_processor, device = 0, points_per_batch = 2)
outputs = generator(imageUrl, points_per_batch = 2)

raw_image = Image.open(imageUrl).convert("RGB")
masks_on_image(raw_image, outputs["masks"])
```

```python
def save_masks_on_images(images, folderpath):
    for image_dict in images:
        image = image_dict['image']
        meta = image_dict['meta']

        imageUrl = meta['file_path']

        generator =  pipeline("mask-generation", model=model, image_processor=processor.image_processor, points_per_batch = 2)
        outputs = generator(imageUrl, points_per_batch = 2)

        raw_image = Image.open(imageUrl).convert("RGB")
        masks_on_image(raw_image, outputs["masks"], show=False, save=True, folderpath=folderpath, filename=meta['file_name'])
        
```

```python
destination_directory_patch = "/home/data/ISIC2019/seg/patch"
destination_directory_no_patch = "/home/data/ISIC2019/seg/no_patch"

print("Start saving patch imgs")
#save_masks_on_images(image_patch, destination_directory_patch)
print("Start saving no patch imgs")
save_masks_on_images(image_no_patch, destination_directory_no_patch)
print("finished saving imgs")
```
