```python
from llama_index.multi_modal_llms.ollama import OllamaMultiModal
import torch
```

```python
seed = 42 # seed doesn't work
torch.manual_seed(seed)
torch.cuda.manual_seed(seed)
mm_model = OllamaMultiModal(model="llava:34b", seed=seed)
```

```python
from pathlib import Path
from llama_index.core import SimpleDirectoryReader
from PIL import Image
import matplotlib.pyplot as plt
import os
import numpy as np
from tqdm import tqdm
```

```python
# load as image documents
image_documents = SimpleDirectoryReader("/home/data/ISIC2019", required_exts=[".jpeg", ".jpg", ".png"], recursive=True, filename_as_id=True).load_data()
```

```python
def metadata_str2dict(string):
    # Split the string into lines
    lines = string.strip().split("\n")
    
    # Create a dictionary to store the key-value pairs
    file_info_dict = {}
    
    # Parse each line and add key-value pairs to the dictionary
    for line in lines:
        key, value = line.split(": ", 1)
        file_info_dict[key] = value
    
    return(file_info_dict)
```

```python
image_patch = []
image_no_patch = []

for image in image_documents:
    metadata = metadata_str2dict(image.get_metadata_str())
    image_path = metadata['file_path']

    folder_name = os.path.basename(os.path.dirname(image_path))
    
    if folder_name == "patch":
        metadata['patch'] = True
        image_patch.append({'image': image, 'meta': metadata})


    elif folder_name == "no_patch":
        metadata['patch'] = False
        image_no_patch.append({'image': image, 'meta': metadata})


    else:
        print("The jpg is not in either the patch or no_patch folder.")
```

```python
num_imgs = 100
num_imgs_per_class = int(np.ceil(num_imgs/2))

# combine images with and without patch
images = []
images.extend(image_patch[:num_imgs_per_class])
images.extend(image_no_patch[:num_imgs_per_class])
```

```python
prompts = [#"The image shows a part of skin with a lesion. Often a patch marks if it is cancer. I want to figure out if this image has such a patch. The patch is usually colourful and has an ellipsoid shape. Do you see a Patch? Please answer with one word, yes or no.",
          #"The image shows a part of skin with a lesion. Sometimes a patch marks if it is cancer. I want to figure out if this image has such a patch. The patch is usually colourful and has an ellipsoid shape. Do you see a Patch? Please answer with one word, yes or no.",
          #"The image shows a lesion and sometimes a coloful ellipsoid patch is placed next to it. Does this image has a patch? Answer with yes or no, with one word.",
          #"You see an image of a part of a skin with a lesion. To mark cancer often a patch is attached next to the lesion. This patch is colorful and ellipsoid. It is not part of the skin. Not evry picture has such a patch. I want to figure out if this image shows a patch next to the lesion. Do you see a patch? Answer with one word of [yes, no].",
          #"You see an image of a part of a skin with a lesion. To mark cancer sometimes a patch is attached next to the lesion. This patch is colorful and ellipsoid. It is not part of the skin. Not evry picture has such a patch. I want to figure out if this image shows a patch next to the lesion. Do you see a patch? Answer with one word of [yes, no].",
          #"I show you images of a part of a skin with a lesion. I want to figure out if a colorful, ellipsoid patch is attached next to the lesion to mark the lesion as malignant. Do you see a colorful, ellipsoid patch attached on the skin? Answer with one of the options [yes, no]", 
          #"I show you images of a part of a skin with a lesion. I want to figure out if a colorful, ellipsoid patch is attached next to the lesion to mark the lesion as malignant. Do you see a colorful, ellipsoid patch attached on the skin? Answer with one word of the options [yes, no]"
          #"The image shows a lesion and sometimes a coloful ellipsoid patch is placed next to it. The patch does not belong to the skin. Does this image has a patch? Answer with yes or no, with one word.",
          #"The image shows part of a skin with a lesion. Sometimes a coloful ellipsoid patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. Do you see a colorful, ellipsoid patch attached on the skin. Answer with one word of the options [yes, no].",
          "The image shows part of a skin with a lesion. Sometimes a coloful ellipsoid patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. Do you see a colorful, ellipsoid patch attached on the skin or not? Answer with one word of the options [yes, no].", 
          #"The image shows a part of skin with a lesion. Often a patch marks if it is cancer. I want to figure out if this image has such a patch. Do you see a patch? Please answer with yes or no. Please answer with one word.",
          #"The image shows part of a skin. On the skin is a lesion that is part of the skin, and maybe also a colorful, ellipsoid patch that is not part of the skin. Can you tell me, if you see a colorful, ellipsoid patch attached on the skin. Reply in one word [yes, no].",
          #"The image shows part of a skin with a lesion. Sometimes a coloful ellipsoid patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. Do you see a colorful, ellipsoid patch attached on the skin or not? Can you do a reasoning but still only reply with one word of the options [yes, no].",
          #"In the image you can see a part of skin with a lesion. The lesion can be malignant or benign. To mark if the lesion is malignant a colorful ellipsoid patch might be attached next to the lesion. The patch does not belong to the skin. Can you see a coloful, ellipsoid patch next to the lesion in this image or not? Answer with one word of the options [yes, no].",
          # "You can get two kind of images. Either you get a part of skin with a lesion. Or a part of skin with a lesion and a colorful, ellipsoid patch attached on the skin. The patch does not belong to the skin. I want to figure out wich of the two options shows this image. Do you see an image with or without colorful ellipsoid patch? Answer with one word of the options [yes, no]",
          #"The image shows part of a skin with a lesion. Sometimes a patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. If there is a patch it is coloful and has an ellipsoid shape. Do you see a colorful, ellipsoid patch attached on the skin or not? Answer with one word of the options [yes, no].",
          #"The image shows part of a skin with a lesion. Sometimes a patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. If there is a patch it is coloful and has an ellipsoid shape. Do you see a colorful, ellipsoid patch attached on the skin or not? \nA. yes there is a patch \nB. no there is no patch \nC. I am not sure if there is a patch. \nAnswer with the option's letter from the given choices directly.",
          #"The image shows part of a skin with a lesion. Sometimes a patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. If there is a patch it is coloful and has an ellipsoid shape. Do you see a colorful, ellipsoid patch attached on the skin or not? Answer with one word of the options [yes, no, unsure].",
          #"The image shows part of a skin with a lesion. Sometimes a patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. The patch color is distinctly different from teh skin color and can vary. If there is a patch it is coloful and has an ellipsoid shape. Do you see a colorful, ellipsoid patch attached on the skin or not? Answer with one word of the options [yes, no].",
          #"The image shows part of a skin with a lesion. Sometimes a coloful ellipsoid patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. The patch has varying colors distinctive from the skin. Do you see a colorful, ellipsoid patch attached on the skin or not, or are you unsure about your answer? Answer with one word of the options [yes, no, unsure].",
          #"The image shows part of a skin with a lesion. Sometimes a coloful ellipsoid patch is placed next to the lesion to mark if it is malignant. The patch does not belong to the skin. Do you see a colorful, ellipsoid patch attached on the skin or not? If you cannot tell if it has a patch answer with unsure. Answer with one word of the options [yes, no, unsure].",
    
          #"Is a patch in the image? Please answer with one word, yes or no.",
          #"Is a patch in the image? Answer with yes or no.",
          #"The image shows a part of skin with a lesion. Often a patch marks if it is cancer. I want to figure out if this image has such a patch. Do you see a patch? Please answer with one word, yes or no.",
          #"Can you see a patch?",
          #"Can you see a patch? Reply with one word, yes or no.",
          #"patch?",
          #"patch? One word reply."
         ]

for prompt in prompts:
    correct_classified = 0
    false_classified = 0
    more_than_one_word = 0
    num_yes_patch = 0
    num_no_no_patch = 0
    num_yes = 0
    num_no = 0
    num_unsure = 0

    print("-----------------------------------------\n")
    print(prompt)
    
    for image_dict in tqdm(images[:num_imgs]):
        image = image_dict['image']
        meta = image_dict['meta']
        
        ## display image
        #imageUrl = meta['file_path']
        #image = Image.open(imageUrl).convert("RGB")
        #plt.figure(figsize=(8, 3))
        #plt.imshow(image)
    
        # prompt 
        prompt= prompt
    
        # response
        image_doc = SimpleDirectoryReader(input_files=[meta['file_path']]).load_data()
    
        complete_response = mm_model.complete(
            prompt=prompt, image_documents=image_doc)
    
        response = str(complete_response).lower().replace(" ", "")
        if response in {"yes", "a", "a."}:
            #print("yes")
            num_yes += 1
            if meta['patch']:
                correct_classified += 1
                num_yes_patch += 1
                #print("correct")
            else:
                false_classified += 1
                #print("false")
    
        elif response in {"no", "b", "b."}:
            #print("no")
            
            num_no += 1
            if meta['patch']:
                false_classified += 1
                #print("false")
            else:
                correct_classified += 1
                num_no_no_patch +=1
                #print("correct")
        elif response in {"unsure", "c", "c."}:
                num_unsure +=1
        else:
            #print("No yes or no!")
            more_than_one_word += 1
            print('Patch: ', meta['patch'])
            print(str(complete_response))

    acc = correct_classified/(num_imgs - more_than_one_word - num_unsure)
    
    print("\nAccuracy: ",acc)
    print("correct classified: ", correct_classified,"\nfalse_classified: ", false_classified, 
          "\nNumber yes: ", num_yes, "\nNumber no: ", num_no, "\nNumber unsure: ", num_unsure, "\nNeither yes or no: ", more_than_one_word,
          "\nNumber True positives: ", num_yes_patch, "\nNumber True negatives: ", num_no_no_patch,
          "\nnum images: ", num_imgs)
    print("\n-----------------------------------------\n\n")
```
