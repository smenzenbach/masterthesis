### Imports

```python
import os
import yaml

from eval_confounder_detection import build_gt_csv, evaluation_metrics, create_evaluation_yaml
```

### Load datasets

```python
# Load the YAML file into a dictionary
dataset_yaml_file = '/workspace/home/data/datasets.yaml'
with open(dataset_yaml_file, 'r') as file:
    datasets = yaml.safe_load(file)

#datasets
```

```python
# get dataset
dataset_key = 'ISIC2019'
dataset = datasets[dataset_key]

dataset_path = f"/workspace/home/data/{dataset['dataset_dir']}"
images_dir_path = f"{dataset_path}/raw"
confounder_dir = datasets[dataset_key]['confounder_dir']
no_confounder_dir = datasets[dataset_key]['no_confounder_dir']
```

# Create ground truth csv file

### Setup output directory


```python
# csv file path
csv_file_gt = os.path.join(dataset_path, 'gt_label.csv')
if not os.path.exists(csv_file_gt):
    build_gt_csv(csv_file_gt, images_dir_path, confounder_dir, no_confounder_dir)
```

# Evaluate confounder detection

```python
# set file path for confounder detection prediction csv file
timestamp = '20240819T093001'

base_dir = f"/workspace/home/pipeline_outputs/{datasets[dataset_key]['dataset_dir']}/{timestamp}"
csv_file_pred = os.path.join(base_dir, 'sgl_pred.csv')

# evaluation metrics
output = evaluation_metrics(csv_file_gt, csv_file_pred)

# create yaml file with evaluation metrics
yaml_file = os.path.join(base_dir, 'confounder_detection_metrics.yaml')
create_evaluation_yaml(yaml_file, output)
        
```
