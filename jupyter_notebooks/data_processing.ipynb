```python
import os
import shutil

def sort_jpg_files(source_dir, dest_dir, file_start, file_end):
    for file_num in range(file_start, file_end+1) :
        source_file = os.path.join(source_dir, 'ISIC_0000' + str(file_num)+ '.jpg')
        
        if os.path.exists(source_file):
            shutil.copy(source_file, dest_dir)

# Example usage
source_directory = "/datasets/ISIC19/raw"
destination_directory_patch = "/home/data/ISIC2019/raw/patch"
destination_directory_no_patch = "/home/data/ISIC2019/raw/no_patch"

if not os.path.exists(destination_directory_patch):
    os.makedirs(destination_directory_patch)

if not os.path.exists(destination_directory_no_patch):
    os.makedirs(destination_directory_no_patch)

# patch images
start_patch = 557
end_patch = 692
sort_jpg_files(source_directory, destination_directory_patch, start_patch, end_patch)

# no patch images
start_no_patch = 352
end_no_patch = 556
sort_jpg_files(source_directory, destination_directory_no_patch, start_no_patch, end_no_patch)
```
